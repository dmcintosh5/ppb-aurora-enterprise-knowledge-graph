```python
import streamlit as st
from agent import PPBAgent

# Page Configuration
st.set_page_config(page_title="PPB HR Assistant", page_icon="🏢")

# --- Session State Management ---
# Ensures the agent is only initialized once and persists across chat turns
if "agent" not in st.session_state:
    with st.spinner("Initializing HR Assistant and connecting to Neo4j..."):
        st.session_state.agent = PPBAgent()

if "messages" not in st.session_state:
    st.session_state.messages = []

# --- UI Layout ---
st.title("Peachtree Pantry & Beverage Co.")
st.subheader("HR Knowledge Assistant")

# Sidebar for System Status
with st.sidebar:
    st.header("System Status")
    st.success("Connected to Neo4j Graph")
    st.info("Model: llama3.2")
    if st.button("Clear Chat History"):
        st.session_state.messages = []
        st.rerun()

# Display Chat History
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# --- Chat Logic ---
if prompt := st.chat_input("Ask about employee policies or the org directory..."):
    # 1. Display User Message
    st.session_state.messages.append({"role": "user", "content": prompt})
    with st.chat_message("user"):
        st.markdown(prompt)

    # 2. Generate and Display Agent Response
    with st.chat_message("assistant"):
        with st.spinner("Searching company documents..."):
            response = st.session_state.agent.ask(prompt)
            st.markdown(response)
    
    st.session_state.messages.append({"role": "assistant", "content": response})
```
